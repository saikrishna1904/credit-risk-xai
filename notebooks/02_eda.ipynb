# Stage 2 – Exploratory data analysis

Default rate, loan grade, purpose, income and other distributions on the cleaned dataset.

Part of **Explainable AI for Credit Risk Prediction**. See the main README for the full pipeline.

```python
# 02_eda.ipynb
# =====================================
# Stage 2: Exploratory Data Analysis (EDA)
# Project: Explainable AI for Credit Risk Prediction (Sai)
# Dataset: accepted_cleaned.csv created in Notebook 1
# =====================================

import polars as pl
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid", palette="muted", font_scale=1.1)

# -----------------------------------------------------
# 1. Load CLEANED dataset (from Notebook 1)
# -----------------------------------------------------
CLEANED_PATH = "../data/accepted_cleaned.csv"  # or "../data/cleanedaccepted.csv"

print(f"🔹 Loading cleaned dataset from: {CLEANED_PATH}")
df = pl.read_csv(CLEANED_PATH, low_memory=True)

print("✅ Cleaned dataset loaded.")
print("Shape:", df.shape)
print("First 15 columns:", df.columns[:15])

# Check that target exists
if "default" not in df.columns:
    raise ValueError("Target column 'default' not found. Check Notebook 1 and cleaned file.")

# -----------------------------------------------------
# 2. Target Variable Distribution
# -----------------------------------------------------
print("\n📊 Target variable distribution (0 = Fully Paid, 1 = Default):")
target_dist = (
    df
    .group_by("default")
    .agg(pl.count().alias("count"))
    .sort("default")
)
print(target_dist)

# Convert to pandas for seaborn
target_pd = target_dist.to_pandas()

plt.figure(figsize=(5,4))
sns.barplot(data=target_pd, x="default", y="count")
plt.title("Target Distribution: Fully Paid vs Default")
plt.xlabel("default (0 = Fully Paid, 1 = Default)")
plt.ylabel("Number of Loans")
plt.tight_layout()
plt.show()

# -----------------------------------------------------
# 3. Identify Numeric & Categorical Columns
# -----------------------------------------------------
print("\n🔹 Detecting numeric vs categorical columns...")

numeric_dtypes = (
    pl.Int8, pl.Int16, pl.Int32, pl.Int64,
    pl.UInt8, pl.UInt16, pl.UInt32, pl.UInt64,
    pl.Float32, pl.Float64
)

numeric_cols = [c for c, dt in zip(df.columns, df.dtypes) if dt in numeric_dtypes and c != "default"]
cat_cols = [c for c, dt in zip(df.columns, df.dtypes) if dt not in numeric_dtypes]

print(f"Numeric columns: {len(numeric_cols)}")
print("Examples:", numeric_cols[:10])
print(f"Categorical columns: {len(cat_cols)}")
print("Examples:", cat_cols[:10])

# -----------------------------------------------------
# 4. Correlation Heatmap (Numeric Features)
# -----------------------------------------------------
if numeric_cols:
    print("\n📈 Computing correlation heatmap on a sample of numeric features...")

    # sample up to 50k rows for speed
    sample_size = min(50_000, df.height)
    df_sample = df.sample(n=sample_size, seed=42)

    corr_pd = df_sample.select(numeric_cols).to_pandas().corr()

    plt.figure(figsize=(12, 9))
    sns.heatmap(corr_pd, cmap="coolwarm", center=0)
    plt.title("Correlation Heatmap (Sampled Numeric Features)")
    plt.tight_layout()
    plt.show()
else:
    print("No numeric columns found for correlation heatmap.")

# -----------------------------------------------------
# 5. Default Rate by Loan Purpose
# -----------------------------------------------------
if "purpose" in df.columns:
    print("\n📊 Default rate by loan purpose...")
    purpose_stats = (
        df.group_by("purpose")
          .agg([
              pl.count().alias("total"),
              pl.col("default").mean().alias("default_rate")
          ])
          .sort("default_rate", descending=True)
    )

    print(purpose_stats)

    purpose_pd = purpose_stats.to_pandas()
    plt.figure(figsize=(10, 6))
    sns.barplot(data=purpose_pd, x="purpose", y="default_rate")
    plt.xticks(rotation=45, ha="right")
    plt.title("Default Rate by Loan Purpose")
    plt.xlabel("Purpose")
    plt.ylabel("Default Rate")
    plt.tight_layout()
    plt.show()
else:
    print("\n⚠️ Column 'purpose' not found – skipping default-by-purpose plot.")

# -----------------------------------------------------
# 6. Default Rate by Grade / Term / Home Ownership / Verification / Emp Length
# -----------------------------------------------------
def plot_default_rate_by(category_col: str, figsize=(8,5)):
    """Helper to compute and plot default rate for a categorical variable."""
    if category_col not in df.columns:
        print(f"⚠️ Column '{category_col}' not found – skipping.")
        return
    print(f"\n📊 Default rate by {category_col}...")

    stats = (
        df.group_by(category_col)
          .agg([
              pl.count().alias("total"),
              pl.col("default").mean().alias("default_rate")
          ])
          .sort("default_rate", descending=True)
    )
    print(stats)

    stats_pd = stats.to_pandas()
    plt.figure(figsize=figsize)
    sns.barplot(data=stats_pd, x=category_col, y="default_rate")
    plt.xticks(rotation=45, ha="right")
    plt.title(f"Default Rate by {category_col}")
    plt.xlabel(category_col)
    plt.ylabel("Default Rate")
    plt.tight_layout()
    plt.show()

# Grade
plot_default_rate_by("grade", figsize=(6,4))

# Subgrade (if not too many)
if "sub_grade" in df.columns:
    # limit to maintainable chart by ordering
    subgrade_stats = (
        df.group_by("sub_grade")
          .agg([
              pl.count().alias("total"),
              pl.col("default").mean().alias("default_rate")
          ])
          .sort("sub_grade")
    )
    print("\n📊 Default rate by sub_grade...")
    print(subgrade_stats)

    sub_pd = subgrade_stats.to_pandas()
    plt.figure(figsize=(10,5))
    sns.barplot(data=sub_pd, x="sub_grade", y="default_rate")
    plt.xticks(rotation=45, ha="right")
    plt.title("Default Rate by Sub-grade")
    plt.xlabel("sub_grade")
    plt.ylabel("Default Rate")
    plt.tight_layout()
    plt.show()

# Term
plot_default_rate_by("term", figsize=(5,4))

# Home ownership
plot_default_rate_by("home_ownership", figsize=(7,4))

# Verification status
plot_default_rate_by("verification_status", figsize=(7,4))

# Employment length
plot_default_rate_by("emp_length", figsize=(9,4))

# -----------------------------------------------------
# 7. Distributions of Key Numeric Features
# -----------------------------------------------------
key_numeric_candidates = [
    "loan_amnt", "funded_amnt", "funded_amnt_inv",
    "int_rate", "installment", "annual_inc", "dti",
    "fico_range_low", "fico_range_high"
]
key_numeric = [c for c in key_numeric_candidates if c in numeric_cols]

print("\n📈 Plotting distributions for key numeric features:", key_numeric)

for col in key_numeric:
    series_pd = df.select(col).to_pandas()[col]

    plt.figure(figsize=(7,4))
    sns.histplot(series_pd.dropna(), bins=50, kde=True)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.tight_layout()
    plt.show()

    # Boxplot by default (to show relationship with target)
    plt.figure(figsize=(6,4))
    tmp_pd = df.select(["default", col]).to_pandas().dropna()
    sns.boxplot(data=tmp_pd, x="default", y=col)
    plt.title(f"{col} vs Default (0/1)")
    plt.xlabel("default")
    plt.ylabel(col)
    plt.tight_layout()
    plt.show()

# -----------------------------------------------------
# 8. Missing Value Summary
# -----------------------------------------------------
print("\n🔹 Missing value summary (post-cleaning)...")

null_counts = df.null_count()  # shape: (1, n_cols)
missing_long = (
    null_counts
    .melt(variable_name="column", value_name="null_count")
    .with_columns(
        (pl.col("null_count") / df.height).alias("missing_pct")
    )
    .sort("missing_pct", descending=True)
)

missing_pd = missing_long.to_pandas()

print(missing_pd.head(20))

plt.figure(figsize=(10,6))
sns.barplot(
    data=missing_pd.head(20),
    x="column",
    y="missing_pct"
)
plt.xticks(rotation=45, ha="right")
plt.title("Top 20 Columns by Missing Percentage")
plt.xlabel("Column")
plt.ylabel("Missing Percentage")
plt.tight_layout()
plt.show()

print("\n✅ EDA completed on cleaned accepted loans dataset.")
```
